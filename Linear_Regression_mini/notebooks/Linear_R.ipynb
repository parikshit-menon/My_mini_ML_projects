```python
import sys
sys.path.append("..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.Linear_R import LinearRegression
```

```python
data = pd.read_csv("../data/houses.csv")

print(data)
```

```python
X = data["size"].to_numpy()
y = data["price"].to_numpy()
```

```python
plt.scatter(X, y)
plt.xlabel("Size")
plt.ylabel("Price")
plt.title("House Prices vs Size")
plt.show()
```

```python
model = LinearRegression()

loss = model.train(X, y, 0.01, 100)
```

```python
plt.plot(loss)
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.title("Training Loss")
plt.show()
```

```python
print("b0:", model.b0)
print("b1:", model.b1)
```

```python
new_sizes = np.array([2.5, 3.5])

new_predictions = model.predict(new_sizes)

print("Predictions:", new_predictions)
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X), color="red")
plt.plot(new_sizes, new_predictions, "go")

plt.xlabel("Size")
plt.ylabel("Price")
plt.title("House Prices vs Size with Regression Line")

plt.show()
```

```python
final_predictions = model.predict(X)

final_mse = model.MSE(y, final_predictions)

print("Final MSE:", final_mse)
```
